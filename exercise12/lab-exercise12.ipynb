```python
%matplotlib inline 
import matplotlib.pyplot as plt 

from sklearn.datasets import load_iris

# iris = load_iris(as_frame=True)
# print(iris.keys())

iris = load_iris()
```

```python
dir(iris)
```

```python
iris.data[0]

# sepal length, sepal width, petal length, petal width
```

```python
iris.target_names[0:5]
```

```python
# split data
from sklearn.model_selection import train_test_split


X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.2)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
# add max_iterator bc total num of iterations reached
model = LogisticRegression(max_iter=1000)

model.fit(X_train, y_train)

model.score(X_test, y_test)
```

```python
iris.target_names[iris.target[22]]
```

```python
prediction = model.predict([iris.data[22]])
iris.target_names[prediction[0]]
```

```python
samples = X_test[:5]

predictions = model.predict(samples)

for i in range(len(samples)):
    print(f"Sample {i+1}: Predicted = {iris.target_names[predictions[i]]}, Actual = {iris.target_names[y_test[i]]}")
```

```python
y_predicted = model.predict(X_test)

from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_predicted)
cm
```

```python
import seaborn as sn 
plt.figure(figsize=(10,7))

sn.heatmap(cm, annot=True)

plt.xlabel('Predicted')
plt.ylabel('Truth')
plt.title('Confusion Matrix')
```

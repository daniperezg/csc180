```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# -----------------------------
# 1️⃣ Create sample dataset
# -----------------------------
np.random.seed(42)  # for reproducibility

# Age between 20 and 75
ages = np.random.randint(20, 76, size=20)

# Income between 80,000 and 300,000
incomes = np.random.randint(80000, 300001, size=20)

df = pd.DataFrame({
    'Age': ages,
    'Income': incomes
})

# -----------------------------
# 2️⃣ Apply Min-Max Scaling
# -----------------------------
mm_scaler = MinMaxScaler()
df['Income_MinMax'] = mm_scaler.fit_transform(df[['Income']])

# -----------------------------
# 3️⃣ Apply Z-score Scaling
# -----------------------------
z_scaler = StandardScaler()
df['Income_Zscore'] = z_scaler.fit_transform(df[['Income']])

# -----------------------------
# 4️⃣ Display Result
# -----------------------------
print(df)
```

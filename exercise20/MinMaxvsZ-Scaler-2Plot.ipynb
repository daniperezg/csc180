```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
# ---------------------------------------------
# 1️⃣ Create a Sample Dataset
# ---------------------------------------------
np.random.seed(42)  # Ensures reproducible random results

# Generate random ages between 20 and 75
ages = np.random.randint(20, 76, size=30)

# Generate random incomes between 80,000 and 300,000
incomes = np.random.randint(80000, 300001, size=30)

# Create a DataFrame
df = pd.DataFrame({
    'Age': ages,
    'Income': incomes
})
print(df.head())
```

```python
# ---------------------------------------------
# 4️⃣ Plot Scaled Data: Scatter Comparison
# ---------------------------------------------
plt.figure(figsize=(12, 5))  # Create a 1x2 layout for plots

# ---- Left: Min-Max Scaling ----
plt.subplot(1, 2, 1)
plt.scatter(df['Age'], df['Income'], color='blue', alpha=0.7)
plt.title('No Scaling (Age vs Income)')
plt.xlabel('Age')
plt.ylabel('Income')
plt.grid(True)


plt.tight_layout()
plt.show()

# ---------------------------------------------
# 5️⃣ Display Scaled Data Sample
# ---------------------------------------------
print(df.head())
```

```python

# ---------------------------------------------
# 2️⃣ Initialize Scalers
# ---------------------------------------------
mm_scaler = MinMaxScaler()     # Scales values to range [0, 1]
z_scaler = StandardScaler()    # Scales values to mean=0 and std=1

# ---------------------------------------------
# 3️⃣ Apply Scaling to Income and Age
# ---------------------------------------------
# Apply Min–Max scaling
df['Income_MinMax'] = mm_scaler.fit_transform(df[['Income']])
df['Age_MinMax'] = mm_scaler.fit_transform(df[['Age']])

# Apply Z-score scaling
df['Income_Zscore'] = z_scaler.fit_transform(df[['Income']])
df['Age_Zscore'] = z_scaler.fit_transform(df[['Age']])

print(df.head())
```

```python
# ---------------------------------------------
# 4️⃣ Plot Scaled Data: Scatter Comparison
# ---------------------------------------------
plt.figure(figsize=(12, 5))  # Create a 1x2 layout for plots

# ---- Left: Min-Max Scaling ----
plt.subplot(1, 2, 1)
plt.scatter(df['Age_MinMax'], df['Income_MinMax'], color='blue', alpha=0.7)
plt.title('Min-Max Scaling (Age vs Income)')
plt.xlabel('Age (Min-Max Scaled)')
plt.ylabel('Income (Min-Max Scaled)')
plt.grid(True)

# ---- Right: Z-Score Scaling ----
plt.subplot(1, 2, 2)
plt.scatter(df['Age_Zscore'], df['Income_Zscore'], color='orange', alpha=0.7)
plt.title('Z-Score Scaling (Age vs Income)')
plt.xlabel('Age (Z-Score Scaled)')
plt.ylabel('Income (Z-Score Scaled)')
plt.grid(True)

plt.tight_layout()
plt.show()

# ---------------------------------------------
# 5️⃣ Display Scaled Data Sample
# ---------------------------------------------
```

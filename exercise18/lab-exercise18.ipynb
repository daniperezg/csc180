```python
import numpy as np 
import matplotlib.pyplot as plt 
import pandas as pd 
import seaborn as sns
```

```python
import warnings 
warnings.filterwarnings('ignore')
```

```python
dataset = pd.read_csv('Melbourne_housing_FULL.csv')

dataset.head()
```

```python
dataset.nunique
```

```python
dataset.shape
```

```python
cols_to_use = ['Suburb', 'Rooms', 'Type', 'Method', 'SellerG', 'Regionname', 'Propertycount', 
               'Distance', 'CouncilArea', 'Bedroom2', 'Bathroom', 'Car', 'Landsize', 'BuildingArea', 'Price']

dataset = dataset[cols_to_use]
dataset.head()
```

```python
dataset.shape
```

```python
dataset.isna().sum()
```

```python
# replace with zero 
cols_to_fill_zero = ['Propertycount', 'Distance', 'Bedroom2', 'Bathroom', 'Car']

dataset[cols_to_fill_zero] = dataset[cols_to_fill_zero].fillna(0)

dataset.isna().sum()
```

```python
# replace with mean 
dataset['Landsize'] = dataset['Landsize'].fillna(dataset.Landsize.mean())

dataset['BuildingArea'] = dataset['BuildingArea'].fillna(dataset.BuildingArea.mean())
```

```python
dataset.isna().sum()
```

```python
# drop small values of 3 
dataset.dropna(inplace=True)

dataset.isna().sum()
```

```python
# get dummies / one hot encoding 
dataset = pd.get_dummies(dataset, drop_first=True)

dataset.head()
```

```python
X = dataset.drop('Price', axis=1)
y = dataset['Price']
```

```python
# split data 
from sklearn.model_selection import train_test_split
train_X, test_X, train_y, test_y = train_test_split(X, y, test_size=0.3, random_state=2)
```

```python
# start with linear regression and observe bad score 
from sklearn.linear_model import LinearRegression
reg = LinearRegression().fit(train_X, train_y)

reg.score(test_X, test_y)
```

```python
# even bad score on train data 
reg.score(train_X, train_y)
```

```python
# L1 Regularization 
from sklearn import linear_model

lasso_reg = linear_model.Lasso(alpha=50, max_iter=100, tol=0.1)

lasso_reg.fit(train_X, train_y)

# as we can see, score increased 
lasso_reg.score(test_X, test_y)
```

```python
# increased on train data too 
lasso_reg.score(train_X, train_y)
```

```python
# L2 Regularization 
from sklearn.linear_model import Ridge

ridge_reg = Ridge(alpha=50, max_iter=100, tol=0.1)

ridge_reg.fit(train_X, train_y)

# just like L1, score increased compared to linear regression 
ridge_reg.score(test_X, test_y)
```

```python
# L2 is slightly below L1
ridge_reg.score(train_X, train_y)
```

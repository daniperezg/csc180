# Using the digts data set, try the following classifiers and find out which one gives the best performance: svm, random forest classifier, logistic regression, gaussian nb, multinomial nb, decision tree clasifier

```python
from sklearn.datasets import load_digits

from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score, KFold
from sklearn.model_selection import GridSearchCV

from sklearn import svm 
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.naive_bayes import MultinomialNB
from sklearn.tree import DecisionTreeClassifier

digits = load_digits()
```

```python
dir(digits)
```

```python
X = digits.data
y = digits.target
```

```python
# split the data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
# kfold to see which model is best 

kfold = KFold(n_splits=5, shuffle=True, random_state=42)

models = {
    "SVM": svm.SVC(),
    "Random Forest": RandomForestClassifier(),
    "Logistic Regression": LogisticRegression(max_iter=2000),
    "GaussianNB": GaussianNB(),
    "MultinomialNB": MultinomialNB(),
    "Decision Tree": DecisionTreeClassifier()
}

for name, model in models.items():
    scores = cross_val_score(model, X, y, cv=kfold)
    print(f"{name}: {scores.mean():.4f}")
```

### As we can see, the models with the highest accuracy are SVM and Random Forest.

```python
# grid search to find optimal hyperparameters 

svm_params = {
    "C": [0.1, 1, 10],
    "gamma": ["scale", "auto"],
    "kernel": ["linear", "rbf"]
}

log_params = {
    "C": [0.1, 1, 10],
    "penalty": ["l2"],
    "solver": ["lbfgs"], 
    "max_iter": [2000]
}

grid = GridSearchCV(svm.SVC(), svm_params, cv=5)
grid.fit(X_train, y_train)

grid2 = GridSearchCV(LogisticRegression(max_iter=2000), log_params, cv=5)
grid2.fit(X_train, y_train)

print("Best SVM Params:", grid.best_params_)
print("Best SVM Score:", grid.best_score_)

print("Best Logistic Params:", grid2.best_params_)
print("Best Logistic Score:", grid2.best_score_)
```

### As we can see, SVM with the parameters of C:10, gamma: scale, and kernel: rbf, was better than Logistic Regression

```python
# test accuracy once again 
best_model = grid.best_estimator_

test_accuracy = best_model.score(X_test, y_test)
print("Test Accuracy:", test_accuracy)
```

### The 99% model score just further proves that the SVM model with the hyperparameters found using gris search is the best model for the digits data set

Using the Iris data set, do k-fold cross validation for Random Forest, SVM, and Logisitc Regression. Which one perform the best?
What happens if you change the Kernel? I.e, do you get 98%? If you use SVC (kernel = poly)

```python
# load models from sklearn and iris data set
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
import numpy as np
from sklearn.datasets import load_iris

iris = load_iris()
dir(iris)
```

```python
# split the data 
from sklearn.model_selection import train_test_split

X = iris.data
y = iris.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
```

```python
# logistic regression 
lr = LogisticRegression()
lr.fit(X_train, y_train)
lr.score(X_test, y_test)
```

```python
# SVC
svm = SVC()
svm.fit(X_train, y_train)
svm.score(X_test, y_test)
```

```python
# random forest 
rf = RandomForestClassifier()
rf.fit(X_train, y_train)
rf.score(X_test, y_test)
```

sample distribution is not fixed and model accuracy changes every time we re-run the split data 

```python
# k-fold 
# from sklearn.model_selection import KFold

# kf = KFold(n_splits=3)
# kf

# stratisfied k-fold ensures that each fold preserves the percentage of samples for each class
from sklearn.model_selection import StratifiedKFold
kf = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
```

```python
# visualize how kfold split works 
# for train_index, test_index in kf.split([0,1,2,3,4,5,6,7,8]):
#     print(train_index, test_index)
```

```python
from sklearn.model_selection import cross_val_score

# define models
models = {
    "Random Forest": RandomForestClassifier(),
    "SVM (RBF)": SVC(kernel='rbf'),
    "Logistic Regression": LogisticRegression()
}

# evaluate each model
for name, model in models.items():
    scores = cross_val_score(model, X, y, cv=kf)

    print(f"{name}: Model Score Average = {np.mean(scores):.4f}")
    print(scores)

# different kernel
poly_svm = SVC(kernel='poly')
poly_scores = cross_val_score(poly_svm, X, y, cv=kf)
print(f"SVM (poly kernel): Model Score Average= {np.mean(poly_scores):.4f}")
print(scores)
```

As we can see, the average accuracy for Random Forest is 0.95, 0.97 for SVM, and 0.97 for Logistic Regression. Changing the SVC kernel didn't make much of a difference here, as they both have the same model scores. All models performed very well with similar scores.

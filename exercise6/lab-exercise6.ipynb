Part 1: Run Gradient Descent algorithm to find m, b, and appropriate learning rate

```python
import math

def gradient_descent(x,y):
    m_curr = b_curr = 0
    iterations = 10000
    n = len(x)
    # changed the learning rate 
    learning_rate = 0.0001

    for i in range(iterations):
        y_predicted = m_curr * x + b_curr
        cost = (1/n) * sum([val**2 for val in (y-y_predicted)])
        md = -(2/n)*sum(x*(y-y_predicted))
        bd = -(2/n)*sum(y-y_predicted)
        m_curr = m_curr - learning_rate * md
        b_curr = b_curr - learning_rate * bd
        print ("m {}, b {}, cost {} iteration {}".format(m_curr,b_curr,cost, i))

x = df['math'].values
y = df['cs'].values

gradient_descent(x,y)
```

Part 2: Use math.is.close() with 1e-20 threshold to compare previous cost to current cost with each iteration

```python
import math

def gradient_descent(x,y):
    m_curr = b_curr = 0
    # added more iterations
    iterations = 100000
    n = len(x)
    learning_rate = 0.0001
    prev_cost = float('inf')

    for i in range(iterations):
        y_predicted = m_curr * x + b_curr
        cost = (1/n) * sum([val**2 for val in (y-y_predicted)])

        # gradients 
        md = -(2/n)*sum(x*(y-y_predicted))
        bd = -(2/n)*sum(y-y_predicted)

        #update 
        m_curr = m_curr - learning_rate * md
        b_curr = b_curr - learning_rate * bd

         # stopping condition
        if math.isclose(prev_cost, cost, rel_tol=1e-20, abs_tol=1e-20):
            print(f"Converged at iteration {i}")
            break

        prev_cost = cost


        print ("m {}, b {}, cost {} iteration {}".format(m_curr,b_curr,cost, i))

    return m_curr, b_curr, cost

x = df['math'].values
y = df['cs'].values

# x = np.array([1, 2, 3])
# y = np.array([2, 4, 6])


m, b, cost = gradient_descent(x,y)
print(f"Gradient Descent: m = {m}, b = {b}, final cost = {cost}")
```

// as we can see, Gradient Descent: m = 1.0262454813586956, b = 1.3121913118610231, final cost = 31.625387089268038
